```sql
-- Business Question 2: How does weather affect demand and trip behavior?
-- Grain: Date + hour + weather condition
-- Metrics: Trip count, average duration, average distance, average fare, average total amount
-- Period: March-May 2026

SELECT
    d.full_date,
    d.day_of_week,
    d.day_name,
    h.hour,
    h.time_of_day,
    w.temperature_band,
    w.precipitation_flag,

    COUNT(*) AS trip_count,
    ROUND(AVG(t.trip_duration_minutes), 2) AS avg_trip_duration_minutes,
    ROUND(AVG(t.trip_distance), 2) AS avg_trip_distance,
    ROUND(AVG(t.fare_amount), 2) AS avg_fare_amount,
    ROUND(AVG(t.total_amount), 2) AS avg_total_amount

FROM nyc_mobility.nyc_gold.fact_taxi_trip AS t

JOIN nyc_mobility.nyc_gold.dim_date AS d
    ON t.pickup_date_key = d.date_key

JOIN nyc_mobility.nyc_gold.dim_hour AS h
    ON t.pickup_hour_key = h.hour_key

LEFT JOIN nyc_mobility.nyc_gold.fact_weather_hourly AS w
    ON t.pickup_date_key = w.date_key
   AND t.pickup_hour_key = w.hour_key

GROUP BY
    d.full_date,
    d.day_of_week,
    d.day_name,
    h.hour,
    h.time_of_day,
    w.temperature_band,
    w.precipitation_flag

ORDER BY
    d.full_date,
    h.hour;
```
