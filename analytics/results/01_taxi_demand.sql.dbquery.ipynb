```sql
-- Business Question 1: Taxi Demand Patterns by Time and Location
-- Grain: Date + hour + pickup zone
-- Metrics: Trip count, total fare, total distance, average trip distance
-- Period: March-May 2026

SELECT
    d.full_date,
    d.day_of_week,
    d.day_name,
    h.hour,
    h.time_of_day,
    h.peak_hour_flag,
    z.borough,
    z.zone,

    COUNT(*) AS trip_count,
    SUM(t.fare_amount) AS total_fare,
    SUM(t.trip_distance) AS total_distance,
    ROUND(AVG(t.trip_distance), 2) AS avg_trip_distance

FROM nyc_mobility.nyc_gold.fact_taxi_trip AS t

JOIN nyc_mobility.nyc_gold.dim_date AS d
    ON t.pickup_date_key = d.date_key

JOIN nyc_mobility.nyc_gold.dim_hour AS h
    ON t.pickup_hour_key = h.hour_key

JOIN nyc_mobility.nyc_gold.dim_zone AS z
    ON t.pickup_location_id = z.location_id

GROUP BY
    d.full_date,
    d.day_of_week,
    d.day_name,
    h.hour,
    h.time_of_day,
    h.peak_hour_flag,
    z.borough,
    z.zone

ORDER BY
    d.full_date,
    h.hour;
```
